```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
plt.rcParams["figure.figsize"] = [7, 4.5]
```

```python
np.random.seed(42)
tamanho = np.random.uniform(30, 200, 150)
preco = 3000 * tamanho + np.random.normal(0, 200, 150) + 5000

df = pd.DataFrame({"tamanho_m2" : tamanho, "preco": preco})
df.to_csv("imoveis.csv", index=False)
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
plt.scatter(df["tamanho_m2"], df["preco"], alpha=0.6, edgecolor="k")
plt.xlabel("Tamanho ao qua ")
plt.ylabel("Preço R$")
plt.title("Tamanho x preço dos imóveis")
plt.show()
```

```python
x = df[["tamanho_m2"]] # 2 [] data frame
y = df["preco"] # 1 [] series
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
print("Treino:", X_train.shape[0], "exemplos")
print("Teste:", X_test.shape[0], "exemplos")
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, Y_train)

print(f"Coeficiente: {model.coef_[0]:.2f}")
print(f"Intercepto: {model.intercept_:.2f}")
```
